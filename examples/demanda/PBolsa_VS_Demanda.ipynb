### 2. Consulta de datos

##### ReadSIMEM

```python
#Consulta de demanda real
id_variable_demandaReal = 'DdaReal'
#Consulta de perdida de energia
id_variable_perdidaEnergia = 'PerdidasEnergia'
#Consulta de precio de bolsa
id_variable_precioBolsa = 'PB_Nal'
#listado de agentes
id_agentes = '972263'

#Fechas de consulta
fecha_inicio = '2025-01-01'
fecha_fin = '2025-12-31'

#Creación de objetos para cada variable
simem_demandaReal = VariableSIMEM(cod_variable=id_variable_demandaReal, start_date=fecha_inicio, end_date=fecha_fin)
simem_perdidaEnergia = VariableSIMEM(cod_variable=id_variable_perdidaEnergia, start_date=fecha_inicio, end_date=fecha_fin)
simem_precioBolsa = VariableSIMEM(cod_variable=id_variable_precioBolsa, start_date=fecha_inicio, end_date=fecha_fin)
simem_agentes = ReadSIMEM(dataset_id=id_agentes, start_date=fecha_inicio, end_date=fecha_fin)
```

## 3.1. Manipulación de datos

```python
# Reinicio de índices y renombrado de columnas
demandaReal = demandaReal.reset_index().rename(columns={'Valor': 'DemandaReal'})
perdidaEnergia = perdidaEnergia.reset_index().rename(columns={'Valor': 'PerdidaEnergia'})
precioBolsa = precioBolsa.reset_index().rename(columns={'Valor': 'PrecioBolsa'})

# Separación de fecha y hora (evitando conversiones repetidas)
for df in [demandaReal, perdidaEnergia, precioBolsa]:
    dt = pd.to_datetime(df['FechaHora'])
    df['Fecha'] = dt.dt.date
    df['Hora'] = dt.dt.hour

# Última versión de agentes
agentes['Fecha'] = pd.to_datetime(agentes['Fecha'])
agentes = (agentes.sort_values('Fecha')
           .drop_duplicates(subset=['CodigoSICAgente'], keep='last')
           [['CodigoSICAgente', 'NombreAgente', 'ActividadAgente']])

# Unión de DataFrames
datos_combinados = (
    demandaReal
    .merge(perdidaEnergia, on=['Fecha', 'Hora', 'CodigoSICAgente', 'TipoMercado'], how='inner')
    .merge(precioBolsa, on=['Fecha', 'Hora'], how='inner')
    .merge(agentes, on='CodigoSICAgente', how='left', suffixes=('_data', '_ag'))
)

# Filtro por comercialización y selección de columnas
datos_combinados = (
    datos_combinados
    .loc[datos_combinados['ActividadAgente'] == 'Comercializador',
         ['Fecha', 'Hora', 'CodigoSICAgente', 'NombreAgente', 'DemandaReal', 'PerdidaEnergia', 'PrecioBolsa', 'Version_x', 'UnidadMedida_x', 'UnidadMedida_y']]
    .groupby(['Fecha', 'Hora', 'CodigoSICAgente', 'NombreAgente', 'Version_x', 'UnidadMedida_x', 'UnidadMedida_y'], as_index=False)
    .agg(DemandaReal=('DemandaReal', 'sum'), PerdidaEnergia=('PerdidaEnergia', 'sum'), PrecioBolsa=('PrecioBolsa', 'mean'))
    .rename(columns={'Version_x': 'Version', 'UnidadMedida_x': 'UnidadMedidaDemanda', 'UnidadMedida_y': 'UnidadMedidaPrecioBolsa'})
)

# Tipo de día
festivos_2025 = holidays.CO(years=2025)
fecha_dt = pd.to_datetime(datos_combinados['Fecha'])
dia_semana = fecha_dt.dt.dayofweek
es_festivo = fecha_dt.isin(festivos_2025)

datos_combinados['TipoDia'] = (
    pd.Series('Día Laboral', index=datos_combinados.index)
    .where(dia_semana != 5, 'Día Sabado')
    .where(dia_semana != 6, 'Día Festivo')
    .where(~es_festivo, 'Día Festivo')
)

# Datos del SIN (agrupación total por hora)
datos_combinados_SIN_agente = (
    datos_combinados
    .groupby(['Fecha', 'Hora', 'Version', 'UnidadMedidaDemanda', 'UnidadMedidaPrecioBolsa', 'TipoDia'], as_index=False)
    .agg(DemandaReal=('DemandaReal', 'sum'), PerdidaEnergia=('PerdidaEnergia', 'sum'), PrecioBolsa=('PrecioBolsa', 'mean'))
    .assign(CodigoSICAgente='SIN', NombreAgente='DEMANDA TOTAL DEL SIN')
    [datos_combinados.columns.tolist()]
)

# Concatenar con datos del SIN
datos_combinados = pd.concat([datos_combinados, datos_combinados_SIN_agente], ignore_index=True)

# Agrupación diaria (solo SIN)
datos_combinados_diarios = (
    datos_combinados_SIN_agente
    .groupby(['Fecha', 'CodigoSICAgente', 'NombreAgente', 'Version', 'TipoDia'], as_index=False)
    .agg(DemandaReal=('DemandaReal', 'sum'), PerdidaEnergia=('PerdidaEnergia', 'sum'), PrecioBolsa=('PrecioBolsa', 'mean'))
)
```

## 3.2. Análisis Gráfico de Datos

```python
#scatterplot diario (sin filtro de tipo día, coloreado por tipo de día)
agentes = sorted(datos_combinados['NombreAgente'].unique().tolist())
tipos_dia = sorted(datos_combinados['TipoDia'].unique().tolist())
def plot_scatter_diario(agente):
    df_filtrado = datos_combinados[
        datos_combinados['NombreAgente'] == agente
    ]
    
    # Agregar por día conservando TipoDia
    df_diario = df_filtrado.groupby(['Fecha', 'TipoDia']).agg(
        DemandaReal=('DemandaReal', 'sum'),
        PrecioBolsa=('PrecioBolsa', 'mean')
    ).reset_index()
    
    # Calcular correlación global
    corr = df_diario['DemandaReal'].corr(df_diario['PrecioBolsa'])
    
    fig = px.scatter(
        df_diario,
        x='DemandaReal',
        y='PrecioBolsa',
        color='TipoDia',
        title=f'Demanda Real vs Precio de Bolsa Diario - Agente: {agente}<br>Corr={corr:.4f}',
        labels={'DemandaReal': 'Demanda Real', 'PrecioBolsa': 'Precio de Bolsa', 'TipoDia': 'Tipo de Día'},
        opacity=0.6
    )
    fig.update_layout(height=500, width=800)
    fig.show()

interact(plot_scatter_diario, 
         agente=Dropdown(options=agentes, description='Agente:'))
```

```python
#scatterplot diario

def plot_scatter_diario(agente, tipo_dia):
    df_filtrado = datos_combinados[
        (datos_combinados['NombreAgente'] == agente) &
        (datos_combinados['TipoDia'] == tipo_dia)
    ]
    
    # Agregar por día
    df_diario = df_filtrado.groupby('Fecha').agg(
        DemandaReal=('DemandaReal', 'sum'),
        PrecioBolsa=('PrecioBolsa', 'mean')
    ).reset_index()
    
    # Calcular correlación
    corr = df_diario['DemandaReal'].corr(df_diario['PrecioBolsa'])
    
    fig = px.scatter(
        df_diario,
        x='DemandaReal',
        y='PrecioBolsa',
        title=f'Demanda Real vs Precio de Bolsa Diario - Agente: {agente} - {tipo_dia}<br>Corr={corr:.4f}',
        labels={'DemandaReal': 'Demanda Real', 'PrecioBolsa': 'Precio de Bolsa'},
        opacity=0.6
    )
    fig.update_layout(height=500, width=800)
    fig.show()

interact(plot_scatter_diario, 
         agente=Dropdown(options=agentes, description='Agente:'),
         tipo_dia=Dropdown(options=tipos_dia, description='Tipo Día:'))
```

```python
def plot_scatter_por_hora(agente, tipo_dia):
    df_filtrado = datos_combinados[
        (datos_combinados['NombreAgente'] == agente) &
        (datos_combinados['TipoDia'] == tipo_dia)
    ]
    
    # Ordenar las horas
    df_filtrado = df_filtrado.copy()
    df_filtrado['Hora'] = df_filtrado['Hora'].astype(int)
    df_filtrado = df_filtrado.sort_values('Hora')
    
    # Calcular correlación por hora
    corr_por_hora = df_filtrado.groupby('Hora').apply(
        lambda x: x['DemandaReal'].corr(x['PrecioBolsa'])
    ).to_dict()
    
    fig = px.scatter(
        df_filtrado,
        x='DemandaReal',
        y='PrecioBolsa',
        facet_col='Hora',
        facet_col_wrap=6,
        title=f'Demanda Real vs Precio de Bolsa por Hora - Agente: {agente} - {tipo_dia}',
        labels={'DemandaReal': 'Demanda Real', 'PrecioBolsa': 'Precio de Bolsa'},
        opacity=0.6,
        category_orders={'Hora': sorted(df_filtrado['Hora'].unique())}
    )
    
    # Actualizar títulos de cada faceta con la correlación
    fig.for_each_annotation(
        lambda a: a.update(text=f"Hora {a.text.split('=')[-1]} Corr={corr_por_hora.get(int(a.text.split('=')[-1]), 0):.3f}")
    )
    
    fig.update_layout(height=800, width=1200)
    fig.show()

interact(plot_scatter_por_hora, 
         agente=Dropdown(options=agentes, description='Agente:'),
         tipo_dia=Dropdown(options=tipos_dia, description='Tipo Día:'))
```

## 4. Análisis adicional

```python
# Variable de interés
nombre_variable = ['Aportes']

# filtrado de variables
idvariable_listado = catalogo_dataset[
    catalogo_dataset['nombreConjuntoDatos']
        .str.lower()
        .str.contains('|'.join([d.lower() for d in nombre_variable]), na=False)
]

idvariable_listado
```

```python
#Consulta de Aportes Hidricos
id_dataset_Aportes = 'BA1C55'


#Fechas de consulta
fecha_inicio = '2025-01-01'
fecha_fin = '2025-12-31'

#Creación de objetos para cada variable
simem_Aportes = ReadSIMEM(dataset_id=id_dataset_Aportes, start_date=fecha_inicio, end_date=fecha_fin)

#Consulta de Aportes Hidricos
aportes = simem_Aportes.main()
```

```python
# Agrupación por fecha sumando los aportes hídricos de energía
aportes = aportes.groupby("Fecha")["AportesHidricosEnergia"].sum()
aportes = pd.DataFrame(aportes).reset_index()
aportes["Fecha"] = pd.to_datetime(aportes["Fecha"])

# Demanda real Diaria del SIN
Datos_diarios = datos_combinados_diarios[datos_combinados_diarios['NombreAgente'] == 'DEMANDA TOTAL DEL SIN'][['Fecha', 'DemandaReal', 'PrecioBolsa', 'TipoDia']].copy()
Datos_diarios["Fecha"] = pd.to_datetime(Datos_diarios["Fecha"])

# Union de informacion de aportes hidricos con demanda real diaria del SIN
Datos_diarios = Datos_diarios.merge(aportes, on='Fecha', how='left')
```

```python
# Scatter interactivo con selección de variables mutuamente excluyentes - Datos_diarios
columnas = ['DemandaReal', 'PrecioBolsa', 'AportesHidricosEnergia']

dd_x = widgets.Dropdown(options=columnas, value='DemandaReal', description='Eje X:')
dd_y = widgets.Dropdown(options=[c for c in columnas if c != 'DemandaReal'], value='PrecioBolsa', description='Eje Y:')

out = Output()
_updating = False

def graficar():
    eje_x = dd_x.value
    eje_y = dd_y.value
    df = Datos_diarios[[eje_x, eje_y]].dropna()
    corr = df[eje_x].corr(df[eje_y])
    fig = px.scatter(
        df,
        x=eje_x,
        y=eje_y,
        title=f'{eje_x} vs {eje_y} - Diario SIN<br>Corr={corr:.4f}',
        labels={eje_x: eje_x, eje_y: eje_y},
        opacity=0.6
    )
    fig.update_layout(height=500, width=800)
    with out:
        out.clear_output(wait=True)
        fig.show()

def on_x_change(change):
    global _updating
    if _updating:
        return
    _updating = True
    dd_y.options = [c for c in columnas if c != dd_x.value]
    _updating = False
    graficar()

def on_y_change(change):
    global _updating
    if _updating:
        return
    _updating = True
    dd_x.options = [c for c in columnas if c != dd_y.value]
    _updating = False
    graficar()

dd_x.observe(on_x_change, names='value')
dd_y.observe(on_y_change, names='value')

graficar()
display(VBox([HBox([dd_x, dd_y]), out]))
```

```python

# Scatter: Aportes Hídricos vs Demanda Real por Tipo de Día
df_plot = Datos_diarios[['AportesHidricosEnergia', 'DemandaReal', 'TipoDia']].dropna()
corr = df_plot['AportesHidricosEnergia'].corr(df_plot['DemandaReal'])

fig = px.scatter(
    df_plot,
    x='AportesHidricosEnergia',
    y='DemandaReal',
    color='TipoDia',
    title=f'Aportes Hídricos vs Demanda Real - Diario SIN<br>Corr={corr:.4f}',
    labels={
        'AportesHidricosEnergia': 'Aportes Hídricos de Energía',
        'DemandaReal': 'Demanda Real',
        'TipoDia': 'Tipo de Día'
    },
    opacity=0.6
)
fig.update_layout(height=500, width=800)
fig.show()
```

```python
#Filtro por demanda Real
demanda_real_MC = demanda[demanda['CodigoVariable'] == 'DdaReal'].copy()

#Agrupacion Diaria por
demanda_real_MC = demanda_real_MC.groupby(['Fecha', 'MercadoComercializacion']).sum().reset_index()[['Fecha', 'MercadoComercializacion', 'Valor']]

#unión de aportes hidricos con demanda real por mercado de comercialización
demanda_real_MC['Fecha'] = pd.to_datetime(demanda_real_MC['Fecha'])
demanda_real_MC_aportes = demanda_real_MC.merge(aportes, on='Fecha', how='left')
demanda_real_MC_aportes
```

```python

# Scatter interactivo: Aportes Hídricos vs Demanda Real filtrado por Mercado de Comercialización
mercados = sorted(demanda_real_MC_aportes['MercadoComercializacion'].dropna().unique().tolist())
dd_mc = widgets.Dropdown(options=['Todos'] + mercados, value='Todos', description='Mercado:')
out_mc = Output()

def graficar_mc(change=None):
    df = demanda_real_MC_aportes.dropna(subset=['AportesHidricosEnergia', 'Valor'])
    if dd_mc.value != 'Todos':
        df = df[df['MercadoComercializacion'] == dd_mc.value]
    corr = df['AportesHidricosEnergia'].corr(df['Valor'])
    fig = px.scatter(
        df,
        x='AportesHidricosEnergia',
        y='Valor',
        color='MercadoComercializacion',
        title=f'Aportes Hídricos vs Demanda Real - {dd_mc.value}<br>Corr={corr:.4f}',
        labels={
            'AportesHidricosEnergia': 'Aportes Hídricos de Energía',
            'Valor': 'Demanda Real',
            'MercadoComercializacion': 'Mercado de Comercialización'
        },
        opacity=0.6
    )
    fig.update_layout(height=500, width=900)
    with out_mc:
        out_mc.clear_output(wait=True)
        fig.show()

dd_mc.observe(graficar_mc, names='value')
graficar_mc()
display(VBox([dd_mc, out_mc]))
```
